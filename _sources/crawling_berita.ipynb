# Crawling Berita

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from IPython.display import display
```

```python
base_urls = [
    "https://www.liputan6.com/news/indeks/2025/09/23",
    "https://www.liputan6.com/news/indeks/2025/09/22"
]

data_artikel = []
```

```python
for base_url in base_urls:
    page = 1
    while True:
        url = f"{base_url}?page={page}" if page > 1 else base_url
        response = requests.get(url)
        if response.status_code != 200:
            print(f"Gagal mengakses {url}")
            break

        soup = BeautifulSoup(response.text, "html.parser")

        # Ambil semua artikel di halaman ini
        articles = soup.find_all("article", class_="articles--rows--item")
        if not articles:
            # Jika sudah tidak ada artikel, hentikan loop
            break

        for article in articles:
            # Judul
            judul_tag = article.find("a", class_="articles--rows--item__title-link")
            judul = judul_tag.get_text(strip=True) if judul_tag else "N/A"

            # Kategori
            kategori_tag = article.find("a", class_="articles--rows--item__category")
            kategori = kategori_tag.get_text(strip=True) if kategori_tag else "N/A"

            # Ringkasan
            ringkasan_tag = article.find("div", class_="articles--rows--item__summary")
            ringkasan = ringkasan_tag.get_text(strip=True) if ringkasan_tag else "N/A"

            # Tanggal dari URL
            tanggal = "-".join(base_url.split("/")[-3:])

            data_artikel.append([judul, tanggal, ringkasan, kategori])

        print(f"✅ Selesai mengambil data dari {url}")
        page += 1  # lanjut ke halaman berikutnya

```

```python
# Simpan ke DataFrame
df = pd.DataFrame(data_artikel, columns=["Judul", "Tanggal", "Ringkasan", "Kategori"])

print("\nDataFrame hasil crawling:")
display(df)

# Opsional: simpan ke CSV
df.to_csv("berita.csv", index=False, encoding="utf-8-sig")
print("Data disimpan ke liputan6_indeks.csv")
```
